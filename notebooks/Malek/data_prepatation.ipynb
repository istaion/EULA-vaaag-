```python
import pandas as pd
import xml.etree.ElementTree as ET
import os
import string
```

```python
string.punctuation
```

```python


def parse_xml_files(rep_path : str = None, file_path : str = None, file_list : list[str] = None) -> list[dict] : 
    
    namespaces = {'tei': 'http://www.tei-c.org/ns/1.0'}
    if file_path : 
        list_xmls = [file_path]
        
    elif rep_path : 
        list_xmls = os.listdir(rep_path)
        
    elif file_list : 
        list_xmls = file_list
        
    result = {}
    for file in list_xmls : 
        tree = ET.parse(f"{rep_path}/{file}")
        root = tree.getroot()

        # Récupérer toutes les balises <w>
        w_elements = root.findall('.//tei:w', namespaces)
        # Construire la liste de dictionnaires
        for w in w_elements : 
            if w.text and w.text not in string.punctuation : 
                result[w.text] = w.get("lemma")
            
    return result
            
            
        
```

```python
files_path = "../../data/xml_data"
dataset = parse_xml_files(rep_path=files_path)
```

```python
len(dataset)
```

```python
df = pd.DataFrame(dataset.items(), columns=["mot", "lemme"])
```

```python
df.to_parquet("../../data/paires_mot_lemme.parquet", index=False)
df.to_csv("../../data/paires_mot_lemme.csv", index=False)
```
